```python
from ultralytics import YOLO
from IPython.display import display
from PIL import Image
import matplotlib.pyplot as plt
import os

os.makedirs("/content/VisionPort", exist_ok=True)

print("VisionPort AI System Started!")
```

```python
!wget -q -O /content/VisionPort/traffic.jpg https://ultralytics.com/images/bus.jpg

image_path = "/content/VisionPort/traffic.jpg"

display(Image.open(image_path))
```

```python
model = YOLO("yolo11n.pt")

print("YOLO Deep Learning Model Loaded Successfully!")
```

```python
results = model.predict(
    source=image_path,
    conf=0.25,
    imgsz=640
)

print("Traffic AI Detection Completed!")
```

```python
detected_image = results[0].plot()

plt.figure(figsize=(14, 8))
plt.imshow(detected_image[:, :, ::-1])
plt.title("VisionPort - AI Traffic Detection")
plt.axis("off")
plt.show()
```

```python
boxes = results[0].boxes

print("VISIONPORT DETECTION REPORT")
print("----------------------------")

for box in boxes:
    class_id = int(box.cls[0])
    confidence = float(box.conf[0])

    object_name = model.names[class_id]

    print(f"{object_name} : {confidence * 100:.2f}%")
```

```python
output_path = "/content/VisionPort/traffic_ai_detected.jpg"

Image.fromarray(detected_image[:, :, ::-1]).save(output_path)

print("Detected image saved successfully!")
display(Image.open(output_path))
```
